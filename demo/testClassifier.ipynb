```python
## Creating a Classifier instance
## Note that the classifier is dedicated to square cropped images
##
from classifTools import Classifier, CROP_SIZE, txt_animalclasses
classifier = Classifier()
```

```python
## Loading the original image
## as a cv2 object
##
import cv2
import numpy as np
imagecv = cv2.imdecode(np.fromfile("testdata/squirrel.JPG", dtype=np.uint8), cv2.IMREAD_UNCHANGED)
```

```python
## Retrieving the cropped image as a PIL object
## using the box coordinates (YOLO format)
## Note that the cropped image is squared, excepted when the animal is at the border of the image
##
from detectTools import cropSquareCVtoPIL
box = [38, 232, 333, 603] # x1, y1, x2, y2
croppedimage = cropSquareCVtoPIL(imagecv, box.copy())
```

```python
## Creating a tensor fro a batch of images
## here a single image
##
import torch
cropped_tensor = torch.ones((1,3,CROP_SIZE,CROP_SIZE))
cropped_tensor[0,:,:,:] =  classifier.preprocessImage(croppedimage)
```

```python
## Prediction on the batch of images
##
scores = classifier.model.predict(cropped_tensor)
# or
# scores = classifier.predictOnBatch(cropped_tensor)
```

```python
import pandas as pd
pd.DataFrame({'species':txt_animalclasses['en'], 'score':scores[0][:]}) 
```
